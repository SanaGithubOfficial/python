# 15. Lamba function in Python
* __A Lamba function is a small anonymous function__
* __A lamba function can take any number of arguments but only have one expression__
* __The expression is evaluated and returned__
* __Lamda expression (or lambda function) essentially blocks of code that can be assigned to variable passed as an arguments or returned from a function call , in Language that support high-ordered function__
* __The main role of lambda function is better described in the scenerious when we employ them anonymously inside another functions__
* __In python , the lambda function can be utilized as an arguments to the higher order functions as an arguments__
* __Lambda function in Python__
   * __Syntax__
   * __Normal Function__
   * __Anonymous Function__
  * __Usage__
    * __Lambda with filters__
    * __Lambda with map()__
    * __Lambda with reduce()__
    * __Lambda with apply()__
* __common errors__
   * __Syntax errors__
   * __type Errors__
