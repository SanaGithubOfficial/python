# Branching (if,elif,else)
* __Decision making is required when we want to execute a code only if a certain condition is satisfied.__
* __The if/elif/else statement is used in Python for decision making.__
* __An else statemnet can be combined with an if statement.__
* __An else statement contains the block of code that executes if the condtiional expression in the first statement resolves to 0 or a False value.__
* __The else statement is an optional statemnet and there could be at most only one else statemnt following if.__
* __The elif statement allows you to check multiple expression for True and execute a block of code as soon as one of the condition evaluates to True.__
* __Similar to the else, the elif statement is optional.__
* __However, unlike else, for which there can be at most one statement, there can be an arbitrary number of elif statemnts following an if.__

```python
age= 5

if age>6:
    print("You can go to primary school")
else:
    print("You are too little ")
```

```python
# 15. Create a simple calculator that performs basic arithmetic operations based on two user input.
# (^,*,/,-,+)
first=float(input("Enter first number"))
operator=input("Enter operator (+,-,*,/,^) :")
second=float(input("Enter Second number"))

if operator == "+":
    result=first+second
    print("result:",result)
elif operator == "-":
    result=first-second
    print("result:",result)
elif operator == "*" :
    result=first*second
    print("result:",result)
elif operator == "/":    
    result=first/second
    print("result:",result)
elif operator == "^":
    result=first^second
    print("result:",result)
else:
    print("Invalid operator")
```
