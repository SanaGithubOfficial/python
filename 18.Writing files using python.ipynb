# 18.Writing files in python
* __To write a text file in Python ,you follow these steps:__
* __1.First, open the text for writing (or appending ) using the open function__
* __2.Second, Write to the text file using write(), or writelines() method.__

### Some Characters and their functions
* __"r" ----------> Open file for reading only. Start reading from the begining of the file. This is default mode.__
* __"rb" ---------> Open file for reading only in binary format. Starts reading from the befinning of the text.__
* __"r+" ----------> Open file for reading and writing only . File pointer placed at beginning of the file.__
* __"w" ---------> Open file writing only. File pointer placed at beginning of the file. Overwrite an existing file and creates a new one if it does not exist.__
* __"wb" ---------> Same as W but opens in binary mode.__
* __"wb++" ---------> Same as wb but also allows reading text from the file.__
* __"a" ---------> Open file for appending . Starts writing at the end of the file . Crete a new file if the file does not exist.__
* __"ab" --------------> Same as a but in binary format. Creates a new file if files does not exists.__
* __"ab+"--------------> Same as ab but also open for reading.__

```python
from reportlab.pdfgen import canvas

filename=input("Enter your filename:")
completeFile="C:\\Users\\sanam\\OneDrive\\Desktop\\"+filename+".txt"
pdfFileName="C:\\Users\\sanam\\OneDrive\\Desktop\\"+filename+".pdf"
with open(completeFile,'w') as f:
    end=int(input("Enter number of lines here:"))
    for i in range(1,end+1):
        userText=input("Enter your text here:")
        f.write(userText)
        f.write("\n")

totalCount=0
vowChar=0
vowList=[]
consChar=1
consList=[]
digit=0
specialChar=0
specialList=[]
with open(completeFile,'r') as x:
    data1=x.read()
    data=data1.replace("\n","")
    for i in data:
        vow=["A","E","I","O","U","a","e","i","o","u"]
        specialCharacter=string.punctuation
        if i in vow:
            vowList.append(i)
            vowChar+=1
        elif  i.isdigit()  :
            digit+=1
        elif i in specialCharacter:
            specialChar+=1
            specialList.append(i)
        else:
            consChar+=1
            consList.append(i)
    totalCount+=1
# print("Total number of characte is:",totalCount)  
# print("list of vowels is:",vowList,"\nTotal number of vowels is :",vowChar)
# print("list of vowels is:",consList,"\nTotal number of vowels is :",consChar)
# print("Total number of digit:",digit)                                                        # ctrl + / to comment multiple lines 

with open(completeFile,'a') as f:
    f.write("--------------Character Analysis-------------------\n")
    f.write(f"Total character in the text is:{totalCount}\n")
    f.write(f"Total number of vowels is: {vowList}={vowList}\n")
    f.write(f"Total number of consonent is: {consList}={consChar}\n")
    f.write(f"Total number of digits is: {digit}\n")
    f.write(f"Total number of special characters is : {specialList}={specialChar}\n")
    f.write("------------------Thanks---------------------------")


c = canvas.Canvas(pdfFileName)
y = 800 # starting from y-position for PDF Text

with open(completeFile,'r') as f:
    lines = f.readlines()
    for line in lines:
        if y<50:
            # if page bottom is reached starts a new page
            c.showPage()
            y = 800
        c.drawString(50,y,line.strip())
        y-=20
c.save()
print(f"PDF file has been generated Successfully at {pdfFileName}")
```
