# 18.Writing files in python
* __To write a text file in Python ,you follow these steps:__
* __1.First, open the text for writing (or appending ) using the open function__
* __2.Second, Write to the text file using write(), or writelines() method.__

### Some Characters and their functions
* __"r" ----------> Open file for reading only. Start reading from the begining of the file. This is default mode.__
* __"rb" ---------> Open file for reading only in binary format. Starts reading from the befinning of the text.__
* __"r+" ----------> Open file for reading and writing only . File pointer placed at beginning of the file.__
* __"w" ---------> Open file writing only. File pointer placed at beginning of the file. Overwrite an existing file and creates a new one if it does not exist.__
* __"wb" ---------> Same as W but opens in binary mode.__
* __"wb++" ---------> Same as wb but also allows reading text from the file.__
* __"a" ---------> Open file for appending . Starts writing at the end of the file . Crete a new file if the file does not exist.__
* __"ab" --------------> Same as a but in binary format. Creates a new file if files does not exists.__
* __"ab+"--------------> Same as ab but also open for reading.__

```python
filename=input("Enter your filename here:")
completeFileName="C:\\Users\\sanam\\OneDrive\\Desktop\\"+filename+".txt"
with open(completeFileName,'w') as f:
    endpoints=int(input("Enter number of lines:"))
    for i in range(1,endpoints+1):
        sentence=input("Enter your text here:")
        f.write(str(i))
        f.write(".")
        f.write(sentence)
        f.write("\n")
```
