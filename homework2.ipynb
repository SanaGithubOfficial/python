```python
#18. Print all prime numbers between 1 and 100.
for num in range(2,101):
    prime = True
    for i in range(2, int(num**0.5)+1):
        if num%i==0:
            prime=False
            break
    if prime:
        print(num,end=" ")
```

### 🔁 Loop with String or Range

```python
#26. Count the number of vowels in a string.
string1="Python is a programming language"
string=string1.replace(" ","")
vowels=["A","E","I","O","U","a","e","i","o","u"]

vowList=[]
vowCount=0
consList=[]
consCount=0
for i in string:
    if i in vowels:
        vowList.append(i)
        vowCount+=1
    else:
        consList.append(i)
        consCount+=1
print("Original string is:",string1)       
print("Vowels in the string is:",vowList,"=",vowCount)
print("Consonent in the string is:",consList,"=",consCount)
```

```python
#27. Check if a string is a palindrome using a loop.
# 1. **madam**
# 2. **racecar**
# 3. **level**
# 4. **radar**
# 5. **civic**
# 6. **refer**
# 7. **deified**
# 8. **rotator**
# 9. **repaper**
# 10. **wow**
# 11. **noon**
# 12. **pop**
# 13. **kayak**
# 14. **stats**
# 15. **tenet**
string=input("enter a palindrome:")
while string != "":
    revs=string.lower()
    if revs[::-1]==revs:
        print("this is palindrome")
    else:
        print("this is not a palindrome")
    string == " "    
    break
```

```python
#for
string1 = input("Enter a palindrome:")
string=string1.lower()
rev=""
for char in string:
    rev=char+rev
if string == rev:
    print("This is palindrome")
else:
    print("This is not a palindrome")
        
```
