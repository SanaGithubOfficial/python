# 19. Decorator in Python
* __Decorator provide a simple syntax for calling higher order functions.__
* __By definition , a decorator is a function that takes another functions and extends their behaviours without modifying them.__
* __Decorator can extremely useful as they allow the extension of an existing functions, without any modifications to the original function source code.__
* __In fact, there are two types of decorator in python. Class Decorator and Function Decorator.__
* __In application , decorators are mainly used in creating middle layer in backend , it perform tasks like token authentication , image compression , validation and many more.__

```python
def decorateCase(val):
    def text():
        function=val()
        textUpper=function.upper()
        return textUpper
    return text

def text1():
    name="reshma"
    return name
    
text1()    
```

```python
decorateCase(text1)
```

```python
def text2() :
    name=input("Enter name here:")
    return name
decorateCase(text2) 
```

```python
@decorateCase
def text3() :
    y=input("Enter name here:")
    print("----------------")
    return y
text3()    
```

```python
@decorateCase
def text4():
    z=input("Enter name:")
    print("-----------")
    return z
text4()    
```

```python
# Write a decorator to take input from user and filter char ,digit
def textFilter(y1):
    def text_digit():
        userInput=input("Enter Your Text Here: ")
        char=''.join([x for x in userInput if x.isalpha()])
        num=''.join([i for i in userInput if i.isdigit()])
        punc='!@#$%^&*()-+_=~`,.'
        specialchar= "".join([y for y in userInput if y in punc])
        return y1(char,num,specialchar)
    return text_digit
@textFilter
def calledFunc(char,num,specialchar):
    print(f"Total Character is:{char}\n")
    print(f"Total numbers is : {num}\n")
    print(f"Total special characters is: {specialchar}")
calledFunc()
```
