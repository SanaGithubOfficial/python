```python
#single inheritance : parent,child
class a:   #parent
    brandname_1 = "Amazon"
    brandname_2 = "Flipkart"
    abc="apple"
class b(a):
    prod_1 = "Online Commerce Store"
    prod_2 = "Online Store"
obj = a()   
obj2 = b()
#print(obj.brandname_1)
#print(obj.brandname_2)
#print(obj2.prod_1)
#print(obj2.prod_2)

obj2=b()
print(obj2.brandname_1)
print(obj2.brandname_2)
print(obj2.prod_1)
print(obj2.prod_2)
print(obj2.abc)
```

```python
#multilevel inheritance: parent,child,grandchild
class a1:  #parent
    brandname1="Amazon"
    brandname2="Flipkart"
class b1(a1):  #child
    prod1="Online Commerce Store"
    prod2="Online Store"
class c1(b1):  #grand child
    popularity1="100"
    popularity2="200"
obj = c1()
print(obj.brandname1)
print(obj.brandname2)
print(obj.popularity1)
print(obj.popularity2)
print(obj.prod1)
print(obj.prod2)
```

```python
# multiple inheritance
class a1: #parent
    brandname1="Amazon"
    brandname2="Flipkart"
class b1: 
    prod1="Online Commerce Store"
    prod2="Online Store"
class c1:
    popularity1="100"
    popularity2="200"
class d(a1,b1,c1):
    x="apple"
    y="Pineapple"
obj = d() 
print(obj.brandname1)
print(obj.brandname2)
print(obj.popularity1)
print(obj.popularity2)
print(obj.prod1)
print(obj.prod2)
print(obj.x)
print(obj.y)
```

```python
#hierarchical inheritance
class a1:
    brandname1="Amazon"
    brandname2="Flipkart"
class b1(a1):   #child of a1
    prod1="Online Commerce Store"
    prod2="Online Store"
class c1(a1):  #child of a1
    popularity1="100"
    popularity2="200"
class d1(a1):
    x="apple"
    y="Pineapple"
obj=b1()
obj1 = c1()   
obj3 = d1()
print(obj3.brandname1)
print(obj3.brandname2)
print(obj3.x)
print(obj3.y)
```

```python
#hybrid inheritance
class a:  #parent
    def roseBlack(self):
        return "Black rose"
class b: #parent
    def roseLavender(self):
        return "Lavender rose"
class c(a):  #child of a
    def purpleRose(self):
        return "purple rose"
class d(b,c):  #child of all
    def childRose(self):
        return "childRose"
obj =d()  
print(obj.childRose())
print(obj.purpleRose())
print(obj.roseLavender())
print(obj.roseBlack())
```

```python
#superInheritance
class Animal():
    def __init__(self,name,breed,food):
        self.name=name
        self.breed=breed
        self.food=food
class Dog(Animal) :
    def __init__(self,name,breed,food,color):
        self.color=color
        super().__init__(name,breed,food)
    #method
    def dogDetails(self):
        print(f"my dog name is : {self.name}")
        print(f"his breed is: {self.breed}")
        print(f"his favorite :{self.food}")
        print(f"My dog is {self.color} in color")
class Horse(Dog):
    def __init__(self,name,breed,food,color,speed):
        self.speed=speed
        super().__init__(name,breed,food,color)
    #method
    def horseDetail(self):
        print(f"my horse name is:{self.name}")
        print(f"his breed is:{self.breed}")
        print(f"his favorite :{self.food}")
        print(f"My horse is {self.color} in color")
        print(f"{self.name} runs {self.speed}")
        
obj = Dog("raftar","labrador","non-veg","white")
obj.dogDetails()
print("-----------------------------------------------")
obj1=Horse("chetak","marwari horse","white","vegitarian","very fast")
obj1.horseDetail()
```

```python
#polymorphism
#overriding
class Animal:
    def speak(self):
        print("Animal speak")
class dog(Animal) :
    def speak(self):
        print("Bark")
class cat(Animal) :
    def speak(self):
        print("meow,meow,meow")
class cow(Animal) :
    def speak(self):
        print("moo")
class goat(Animal):
    def speak(self):
        print("Bleat")
ob=cat()
ob.speak()
```

```python
#encapsulation
class encapsulation:
    def __init__(self,value):
        self.value=value
    def displayValue(self):
        return self.value
    def updateValue(self,newValue):
        self.value=newValue
obj1 = encapsulation(10) 
obj1.displayValue()
```

```python
obj1.displayValue()
```

```python
obj1.updateValue(20)
```

```python
#data abstraction
from abc import ABC,abstractmethod
class MainBranch(ABC):
    def database1(self):
        print("Database Connected")

    @abstractmethod    #decorator
    def security(self):
        pass
    @abstractmethod
    def security1(self):
        pass
    @abstractmethod
    def abc(self):
        pass
        
obj1= MainBranch()
obj1.database1()
```

```python
class PuneBranch(MainBranch):
    def database2(self):
        print('Database connected of Pune branch')
    
    def security(self):
        pass
    
    def security1(self):
        pass
    
    def abc(self):
        pass
obj2=PuneBranch()   
obj2.database2()
obj2.database1()
```
