```python
# Logic Operations in Set

# Union
# Intersection
# Difference
# Symmetric Difference
set1 = {1,2,3,4,5,6}
set2 = {5,6,7,8,9,12,50}

print(set1)
print(set2)
```

```python
# {1,2,3,4,5,6,7,8,9,12,50}

# union
print(set1.union(set2))
```

```python
# Difference

# a-b 
# b-a
set1 = {1,2,3,4,5,6}
set2 = {5,6,7,8,9,12,50}
print(set1) #a
print(set2) #b
```

```python
#is subset and superset
a={1,2,3}
b=set([1,2,3,4])
print(a,b)
```

```python
# Subset:
# A set A is considered a subset of set B if all elements of set A are also present in set B. 
# In other words, every member of the subset is also a member of the superset. 

# Superset:
# A set B is a superset of set A if set B contains all elements of set A. 
# A superset can contain additional elements not present in the subset.
```

```python
setA={1,2,3}
setB={1,2,3,4}
print(setA.issubset(setB))
print(setB.issubset(setA))
```

```python
setA={1,2,3}
setB={1,2,3,4}
print(setA.issuperset(setB))
print(setB.issuperset(setA))
```
