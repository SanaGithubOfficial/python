# 12. exception handling:
* __An exception handling is an event occurs during the execution of a program that disrupt the normal flow of the program instructions.__ 
* __In general when a python script encounters that it cannot cope with it it raises an exceptions.__
* __An exception is a python object that represents an error.__
* __An exception is a python script raises an exceptions it must either handle the exception immediately otherwise it terminates and quits.__
* __If you have some suspicious code that may raise an exceptions you can defend you program by placing the suspicious code in a try:block.__
* __After that try:block include an except:statement followed by a block of code which hansles the problems as elegently as possible.__
* __Some common exceptions:__
* __ZeroDivisionError__
* __NameError__
* __ValueError__
* __IOError(inputOutputError)__
* __EOFError(End of File Error)__
* __IdentationError__

```python
# wap to take some string from user and display string starts from 'A'
listVal=[]
charA=[]
totalchar=0
totalcharA=0
while True:
    userVal=input("Enter user Value:")
    if userVal == 'stop':
        break
    listVal.append(userVal)
    totalchar+=1
    if userVal[0] == 'a' or userVal[0] == 'A':
        charA.append(userVal)
        totalcharA+=1
print("list of values is:",listVal,"=",totalchar,"\nlist of values starts with character A:",charA,"=",totalcharA)        
```
