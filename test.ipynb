```python
# Q10. Create a function to count the number of vowels and consonants in a string.
string1 = "The quick brown fox jumps over the lazy dog"
string=string1.replace(" ","")
vowels="AEIOUaeiou"

def vowcons():
    vowChar=0
    vowlist=[]
    consChar=0
    conslist=[]
    for i in string:
        if i in vowels:
            vowlist.append(i)
            vowChar+=1
        else:
            consChar+=1
            conslist.append(i)
    print(f"total vowels:{vowlist}={vowChar}")
    print(f"total consonent:{conslist}={consChar}")
vowcons()
```
