```python
from datetime import datetime, timedelta, UTC
def getWeather(APIkey,city,countrycode):
    url=f"https://api.openweathermap.org/data/2.5/weather?q={city},{countrycode}&appid={APIkey}"
    print(url)

    response = requests.get(url)
    print(response)

    if response.status_code == 200:
        data=response.json()
        print(data)

        print("-----------Type-------------",type(data))

        cityName = data["name"] +","+data["sys"]["country"]
        print(cityName)

        temp=data["main"]["temp"] 
        x1=temp-273.15
        print("Temperature:",temp,"˚C")

        feel=data["main"]["feels_like"]
        x2=feel-273.15
        print("Feels like:",feel,"˚C")

        dateTime=data["timezone"]
        utc_now=datetime.now(UTC)
        localtime= utc_now+timedelta(seconds=data["timezone"])
        #print("Local time in ",data["name"],"is:",localtime.strftime('%Y-%m-%d %H:%M:%S %p'))

        print("Local Time in", data["name"], "is:", localtime.strftime('%Y-%m-%d %H:%M:%S %p'))


        
getWeather("7bdb30c7c674e74a7c684d55516cb319","London","GB")        
```
