```python
#4. Define a function to find the *sum of all digits* of a number.
def sumdigit(num):
    num=abs(num)
    strNum=str(num)
    sumDigit=0
    for digit in strNum:
        sumDigit += int(digit)
    return sumDigit 
    
obj1=sumdigit(123424) 
obj1
```

```python
#9. Create a function to check if a number is *Armstrong number*.
def armstrongnum():
    num=int(input("Enter a number:"))
    org=num
    sum=0
    n=len(str(num))
    while num>0:
        digit=num%10
        sum += digit ** n
        num //= 10
    if sum == org:
        print(f"{org} is armstrong number.")
    else:
        print(f"{org} is not armstrong number.")
obj=armstrongnum()  
obj
```
