# 20.Generator in Python
* __Python generators are function that returns the traversable object and a simple way of creating iterators.__
* __It traverses all the items at once.__
* __The generator also can be expressed in a syntax similar to the list.__
* __There is a lot of complexity in creating iteration in Python. It is required to implement iter() and next() methods to keep track of internal states.__
* __It is a lengthy process to create iterators; It raise stopIteration exceptions.__
* __It pauses the function execution by saving all states and yields to the caller.__
* __Later it resumes the execution when a succesive function is called.__

```python
def checkVal(x=20):
    if x%2==0:
        return x
checkVal(20)        
```

```python
help(checkVal)
```

```python
def checkVal(x=20):
    if x%2==0:
        yield x
checkVal(20)        
```

```python
list =[]

for x in range(20):
    list.append(x)
print(list)    
```

```python
def displayVal():
    for i in range(20):
        yield i
displayVal()        
```

```python
list2=[]
for i in displayVal():
    list2.append(i)
print(list2)    
```

```python
def function():
    for i in range(20):
        if i%2==0:
            yield i
function()            
```

```python
listEven=[]
for j in function():
    listEven.append(j)
print(listEven)    
```
