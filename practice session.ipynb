```python
## 3. List Operations

# Create a list of 5 numbers entered by the user. Then:

# Add a number to the list

# Remove the second element

# Sort the list in descending order

# Print the final list
numbers=[]
for i in range(5):
    num=int(input("Enter a number:"))
    numbers.append(num)
    #add number
    numbers.append(20)
    #remove 2nd element
    if len(numbers) > 1:
        numbers.pop(1)
    #sort descending
    numbers.sort(reverse=True)
print("Final list",numbers)
        
```
