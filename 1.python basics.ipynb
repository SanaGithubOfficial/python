# Data Typecasting

```python
x=10                       #how do you consider a datatype as you want
print(x,type(x))

string=str(x)
print(string,type(string))

float1=float(x)
print(float1,type(float1))
```

```python
#boolean to string="True", integer=1, float=1.0
bool1=True

bool2=False

print(bool1,type(bool1))

print(bool2,type(bool2))

bool3=str(bool1),(bool2)
print(bool3,type(bool3))

bool4=str(bool1),(bool2)
print(bool4,type(bool4))



```
