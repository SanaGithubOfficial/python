# 9. Function in Python
* __In Python, a funciton is group of related statemnets that performs a specific task.__
* __Functions helps to break our program into smaller and modular chunks. As per our program grows larger and alrger, functions make it more organized and manageable.__
* __Furthermore, it avoids repetitions and make code reusable.__
* __There are two types of functions.__
* __1.Pre-defined function__
* __2.User-defined function.__
* __In Python a function is defined using the def keyword followed by the function name and parentheses().__
* __Keyword def that marks to start the funciton header.__
* __A funcitons name to uniquely identify the function.__
* __Funciton naming follows the same rules of writing identifiers in Python.__
* __Parameters(arguments) through which we pass values to a funciton. They are optional.__
* __A colon(:) to mark the end of the function header.__
* __Optional documentation string(docstring) to describe what the function does.__
* __One or more valid statements that marks up the functiion body.__
* __Statements must have the same identation level(usually 4 spaces).__
* __An Optional return statemtns that return a value from the function.__
* __A function also have:__
   * __Function with retun value and function without return value.__
   * __Function with parameter is called parameterized function.__
   * __Function without parameter is called Nullable function.__

* __the scope of a variable is the part of the program to which that variable is accessible__
