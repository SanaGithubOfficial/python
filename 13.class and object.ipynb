```python
# wap to count vowel and their consonent in users name using class and object.
class countAlpha():
    def __init__(self):
        self.name1=input("Enter your name:")
        self.name=self.name1.replace(' ',"")
    def countvowchar(self):
        if self.name.isalpha():
            #print(self.name)
            vowchar=[]
            conschar=[]
            for x in range(len(self.name)):
                vow="AEIOUaeiou"
                if self.name[x] in vow:
                    vowchar.append(self.name[x])
                else:
                    conschar.append(self.name[x])
            print("vowel characters is:",vowchar,"\nConsonent characters is:",conschar)     
        else:
            print(f"{self.name} is not a string type")
obj = countAlpha()    
obj.countvowchar()
```

```python
#6.find the area of right-angle triangle and their parameter 
import math
class rightangleTriangle:
    def __init__(self):
        self.base= int(input("Enter base:"))
        self.height= int(input("Enter height:"))
    def areaTriangle(self):
        area=(1/2)*self.base*self.height
        print("Area of right-angle triangle is:",area,"square units")
        hypotenuse=math.sqrt((self.base**2)+(self.height**2))
        print("Hypotenuse is:",hypotenuse)
        perimeter=self.base+self.height+hypotenuse
        print("Perimeter of right-angle triangle is:",perimeter,"units")
obj = rightangleTriangle()  
obj.areaTriangle()
```

```python
#8.find the area of isosclaces triangle and their parameter
#if you know only base and equal sides
import math
class isosceles:
    def __init__(self):
        self.equalsides=int(input("Enter two equal sides:"))
        self.base=int(input("Enter base:"))
    def areaofisosceles(self):
        area=(self.base/4)*(math.sqrt((4*(self.equalsides*self.equalsides))-(self.base*self.base)))
        print("Area of isosceles using 2 equal sides and base is:",area,"square units")
        perimeter=2*self.equalsides+self.base
        print("Perimeter of isosceles is:",perimeter,"units")
    
    def areawithheight(self):
        self.height=int(input("Enter height if you know:"))
        area=(1/2)*self.base*self.height
        print("area with using hight and base is :",area,"square units")
obj = isosceles()   
obj.areaofisosceles()
obj.areawithheight()        
```

```python
#10.find the area of Trapezium and their parameter
#area = ((1/2)*(a+b)*h) leg= sqrt(b*b-a*a)/2+h*h
#perimeter = a+b+c+d or a+b+2*leg
class trapezium:
    def __init__(self):
        self.side1=int(input("Enter side1:"))
        self.side2=int(input("Enter side2:"))
        self.height=int(input("Enter height:"))
    def areaTrapezium(self):
        area=(1/2)*(self.side1+self.side2)*self.height
        print("area of trapezium is:",area,"cm2")
        leg=math.sqrt((((self.side2-self.side1)/2)**2)+self.height**2)
        perimeterTrapezium=self.side1+self.side2+(2*leg)
        print("perimeter of trapezium is:",perimeterTrapezium,"units")
a= trapezium()  
a.areaTrapezium()
```

```python
#11.find the area of cube and their parimeter
class cube:
    def __init__(self):
        self.side=int(input("Enter side:"))
    def areaCube(self):
        area=6*(self.side*self.side)
        print("Area of cube is:",area,"suare units")
        perimeter=4*self.side
        print("perimeter of one face is:",perimeter,"units")
        totalEdgeofLength=12*self.side
        print("Total edge of length is:",totalEdgeofLength,"units")
obj = cube()  
obj.areaCube()
```

```python
#12.find the area of cuboid and their parameter 
class cuboid:
    def __init__(self):
        self.length=int(input("Enter length:"))
        self.width=int(input("Enter width:"))
        self.height=int(input("Enter height:"))
    def areaCuboid(self):
        area=2*((self.length*self.width)+(self.length*self.height)+(self.width*self.height))
        print("Area of cuboid is:",area,"square units")
        perimeter=2*(self.length+self.height)
        print("perimeter of front face is:",perimeter,"units")
        edgelength=4*(self.length+self.width+self.height)
        print("total edge length is:",edgelength,"units")
obj = cuboid()   
obj.areaCuboid()
```

```python
#14.find the area of hemisphere and perimeter
class hemisphere:
    def __init__(self):
        self.radius=int(input("Enter radius:"))
    def areaSphere(self) :
        curvedSurfaceArea= 2*math.pi*(self.radius*self.radius)
        print("curved surface area of sphere is:",curvedSurfaceArea,"cm2")
        totalSurfaceArea=3*math.pi*(self.radius*self.radius)
        print("total surface area of sphere is:",totalSurfaceArea,"cm2")
        perimeter=2*math.pi*self.radius
        print("perimeter of sphere is:",perimeter,"cm")
obj = hemisphere()    
obj.areaSphere()
```

```python
#15.find the area of cylinder and their volume
class cylinder():
    def __init__(self):
        self.radius=int(input("Enter radius:"))
        self.height=int(input("Enter height:"))
    def areaCylinder(self):
        curvedSurfaceArea= 2*math.pi*self.radius*self.height
        print("curved surface area of cylinder is:",curvedSurfaceArea,"cm2")
        totalSurfaceArea=(2*math.pi*self.radius*self.height)+2*math.pi*(self.radius*self.radius)
        print("total surface area of cylinder is:",totalSurfaceArea,"cm2")
        volume=math.pi*(self.radius*self.radius)*self.height
        print("volume of cylinder is:",volume,"cm3")
obj = cylinder()   
obj.areaCylinder()
```

```python
#16.find the area of cone and their volume
#curved surface area=pi*r(radius)*l(slant height)
#total surface area = pi*r(radius)(l(slant height)+r(radius))
#volume = 1/3*pi*(r(radius)**2)*h(height)

class cone:
    def __init__(self):
        self.radius=int(input("Enter radius:"))
        self.height=int(input("Enter height:"))
        
    def areaCone(self):
        slantHeight=math.sqrt((self.radius*self.radius)+(self.height*self.height))
        print("slant height:",slantHeight,"cm")
        curvedSurfaceArea=math.pi*self.radius*slantHeight
        print("curved surface area of cone is:",curvedSurfaceArea,"cm2")
        totalSurfaceArea=math.pi*self.radius*(slantHeight+self.radius)
        print("total surface area of cone is:",totalSurfaceArea,"cm2")
        volume=(1/3)*math.pi*(self.radius*self.radius)*self.height
        print("volume of cone is:",volume,"cm3")
obj = cone()   
obj.areaCone()
```
