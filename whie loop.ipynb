```python
# 1-10 sum using while loop, even and odd

# 1-10 = 55 , even = 30, odd=25

i=1
evenSum=0
oddSum=0

listEven=[]
listOdd=[]

sqrEven=[]
sqrSum=0

sqrOdd=[]
sqroddsum=0
while(i<=10):
    if i%2==0:
        #print(i)
        listEven.append(i)
        evenSum+=i

        y=i**2
        sqrEven.append(y)
        sqrSum+=y
    else:
        #print(i)
        listOdd.append(i)
        oddSum+=i
        z=i**2
        sqrOdd.append(z)
        sqroddsum+=z
    i+=1
print("list of even number is:",listEven,"=",evenSum,"\nlist of odd number is:",listOdd,"=",oddSum)   
print("list of even number square is:",sqrEven,"=",sqrSum,"\nlist of odd number square is:",sqrOdd,"=",sqroddsum)
```
