# 16.List Comprehension
* __List Comprehension in python is an easy and compact syntax for creating a list from a string of another list__
* __It is very consise way to create a new list by performing an operating on each item in the existing list.__
* __List comprehension is considerably faster than precessing a list using the for loop.__

```python
# 2. Use a lambda function with map() to add 5 to each number in the list:
nums = [1, 2, 3, 4]
def add5(a,b):
    for i in nums:
        i+=5
        print(i)
add5()        
```

```python
# 3. Use filter() with a lambda to find even numbers from a list:
nums1 = [5, 8, 11, 14, 17, 20]
def even():
    for i in nums1:
        if i%2==0:
            print(i)
even()            
```

```python
list(filter(even,nums1))
```

```python
# 4. Use reduce() with a lambda to compute the product of all elements in a list:
nums = [2, 3, 4, 5]
#(Hint: from functools import reduce)
from functools import reduce
def product(a,b):
    return a*b
product(2,3)    
```

```python
# 17. Given a list of words, extract words that contain the letter ‘e’:
words = ['tree', 'book', 'python', 'sky']
for i in range(len(words)):
    if 'e' in words[i]:
        print(words[i])
```

```python
[words[i] for i in range(len(words)) if 'e' in words[i]]
```

```python
# 18. Create a list of all 3-digit numbers that are palindromes using list comprehension.
for num in range(100,1000):
    if str(num)==str(num)[::-1]:
        print(num,end=' ')
```

```python
print([num for num in range(100,1000) if (str(num)==str(num)[::-1])])
```

```python
# 19. From a list of numbers, replace odd numbers with 'ODD' and keep even numbers as is:
nums = [1, 2, 3, 4, 5, 6]
oddList=[]
for i in nums:
    if i%2!=0:
        oddList.append("ODD")
        #print(i)
print(oddList)        
```

```python
["ODD" for i in nums if i%2!=0 ]
```

```python
["ODD" if i%2!=0 else i for i in nums]
```

```python
# 20. Using nested list comprehension, generate a 3x3 identity matrix.
matrix=[[1 if i==j else 0 for j in range(3)] for i in range(3)]
print(matrix)
```
