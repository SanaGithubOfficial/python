# Loops in Python:
* __A for loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).__
* __This is less like the for keyword in other programming languages, and works more like an iterator method as found in other object-oriented programming langueges.__
* __With the for loop we can execute a set of statements, once for each item in a list, tuple, set etc.__
* __The for loop does not require an indexing variable to set beforehand .__
* __With the while loop we can execute a set of statements as long as a condition is true.__
* __Note: remember to increment i,or else the loop will continue forever .__
* __The while loop requires relevant variables to be ready, in this example we need to define an indexing variable, i, which we set to 1__

## range() function:
* __It is helpful to think of the range object as an ordered list.__
* __To loop through a set of code a specified number of times ,we can use the range() function__
* __The range() function returns a sequence of numbers, starting from 0 by default, and incremens by 1 (by default) , and ends at a specified number.__

```python
# wap to input user's name and count total number of vowel character and consonenet

name=input("Enter your name here: ")
name1=name.lower()
vow=0
vowChar=[]

cons=0
consChar=[]

totalChar=0
for i in name:
    if i in ["a","e","i","o","u"]:
        vowChar.append(i)
        vow+=1
    else:
        cons+=1
        consChar.append(i)
    totalChar+=1
print(vowChar,"=",vow) 
print(consChar,"=",cons)
print("Total character is:",totalChar)
```
