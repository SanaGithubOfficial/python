# 6. Dictionary in Python
* __Dictioanries are used to sotre data values in key:values pair.__

* __A dictionary is a collection which is ordered, changeable or mutable and do not allow duplcates.__

* __Dictionary items are ordered, changeable, and do not allow duplicate.__

* __Dictionary items are presented in key:value pairs, and can be referred to by using the key name.__

* __Dictionries are changeable, meaning that we can change, add or remove items after the dictionary has been created.__

* __Dictionary can not have two items with the same key.__

* __A dictionary can nested and can contains another dictionary.__

* __"Keys"------------"Values"__
* __'name'-----------"John"__
* __'age'------------"25"__
* __'city'------------"New York"__
* __'email'----------"bob@web.com"__
